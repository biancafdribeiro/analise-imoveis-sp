# 🏙️ Simulador de Preço — Imóveis SP

Notebook de aplicação autossuficiente. Carrega o modelo salvo pelo `Modelagem.ipynb` e exibe um simulador interativo.

> **Pré-requisito:** execute `Modelagem.ipynb` até a célula de `joblib.dump` para gerar os arquivos `.pkl` em `../data/`.

```python
import pandas as pd
import numpy as np
import joblib
import ipywidgets as widgets
from IPython.display import display, clear_output
import os
```

## 1 · Carrega modelo e dados

```python
rf             = joblib.load('../data/random_forest_model.pkl')
colunas_modelo = joblib.load('../data/colunas_modelo.pkl')

df = pd.read_csv('../data/imoveis_sp_limpo.csv', encoding='utf-8-sig')
df = df.dropna()
df = df[(df['preco_venda'] >= df['preco_venda'].quantile(0.01)) &
        (df['preco_venda'] <= df['preco_venda'].quantile(0.99))]

bairros = sorted(df['bairro'].unique())
tipos   = sorted(df['tipo_imovel'].unique())

print(f'Modelo carregado | {len(colunas_modelo)} features')
print(f'Base: {len(df):,} imóveis | {len(bairros)} bairros | {len(tipos)} tipos')
```

## 2 · Simulador interativo

Ajuste os controles e clique em **Salvar** para exportar a previsão com um imóvel real comparável.

```python
# ── Widgets de entrada ───────────────────────────────────────────
style = {'description_width': '130px'}
w400  = widgets.Layout(width='420px')
w500  = widgets.Layout(width='520px')

w_bairro = widgets.Dropdown(
    options=bairros,
    value='Moema' if 'Moema' in bairros else bairros[0],
    description='📍 Bairro:',
    style=style, layout=w400
)
w_tipo = widgets.Dropdown(
    options=tipos,
    value='Apartamento' if 'Apartamento' in tipos else tipos[0],
    description='🏠 Tipo:',
    style=style, layout=w400
)
w_area = widgets.IntSlider(
    min=20, max=750, step=10, value=100,
    description='📐 Área (m²):',
    style=style, layout=w500
)
w_quartos = widgets.IntSlider(
    min=0, max=10, value=2,
    description='🛏️ Quartos:',
    style=style, layout=w400
)
w_banheiros = widgets.IntSlider(
    min=0, max=10, value=2,
    description='🚿 Banheiros:',
    style=style, layout=w400
)
w_suites = widgets.IntSlider(
    min=0, max=10, value=1,
    description='🛁 Suítes:',
    style=style, layout=w400
)
w_vagas = widgets.IntSlider(
    min=0, max=10, value=1,
    description='🚗 Vagas:',
    style=style, layout=w400
)

out       = widgets.Output()
out_salvo = widgets.Output()

# Guarda o estado atual da previsão para o botão de salvar
estado_atual = {}

# ── Função: busca imóvel real mais parecido ───────────────────────
def buscar_similar(bairro, tipo, area):
    """Busca no dataset o imóvel com mesmo bairro+tipo e área mais próxima."""
    sub = df[(df['bairro'] == bairro) & (df['tipo_imovel'] == tipo)].copy()
    if len(sub) == 0:
        # fallback: só mesmo bairro
        sub = df[df['bairro'] == bairro].copy()
    if len(sub) == 0:
        return None
    sub['dist_area'] = (sub['area_util'] - area).abs()
    return sub.loc[sub['dist_area'].idxmin()]

# ── Função de previsão ────────────────────────────────────────────
def prever(change=None):
    novo = pd.DataFrame([{
        'bairro':        w_bairro.value,
        'tipo_imovel':   w_tipo.value,
        'area_util':     float(w_area.value),
        'quartos':       float(w_quartos.value),
        'banheiros':     float(w_banheiros.value),
        'suites':        float(w_suites.value),
        'vagas_garagem': float(w_vagas.value),
    }])

    novo_enc = pd.get_dummies(novo, columns=['bairro', 'tipo_imovel'])
    novo_enc = novo_enc.reindex(columns=colunas_modelo, fill_value=0)

    # Sem log — previsão direta em R$
    preco = rf.predict(novo_enc)[0]
    low   = preco * 0.85
    high  = preco * 1.15

    # Mediana do bairro+tipo
    filtro     = (df['bairro'] == w_bairro.value) & (df['tipo_imovel'] == w_tipo.value)
    sub        = df[filtro]['preco_venda']
    med_bairro = sub.median() if len(sub) >= 3 else df[df['bairro'] == w_bairro.value]['preco_venda'].median()
    n_imoveis  = len(sub) if len(sub) >= 3 else len(df[df['bairro'] == w_bairro.value])
    delta      = (preco - med_bairro) / med_bairro * 100
    sinal      = '▲' if delta > 0 else '▼'

    # Imóvel real mais parecido — extrai valores antes do with out
    similar = buscar_similar(w_bairro.value, w_tipo.value, w_area.value)
    if similar is not None:
        sim_bairro  = str(similar['bairro'])
        sim_tipo    = str(similar['tipo_imovel'])
        sim_area    = float(similar['area_util'])
        sim_quartos = int(similar['quartos'])
        sim_vagas   = int(similar['vagas_garagem'])
        sim_preco   = float(similar['preco_venda'])
        sim_diff    = (preco - sim_preco) / sim_preco * 100
        sim_sinal   = '▲' if sim_diff > 0 else '▼'
    else:
        sim_bairro = sim_tipo = sim_area = sim_quartos = sim_vagas = sim_preco = None

    # Salva estado para o botão
    estado_atual.update({
        'bairro':        w_bairro.value,
        'tipo_imovel':   w_tipo.value,
        'area_util':     w_area.value,
        'quartos':       w_quartos.value,
        'banheiros':     w_banheiros.value,
        'suites':        w_suites.value,
        'vagas_garagem': w_vagas.value,
        'preco_previsto':preco,
        'intervalo_min': low,
        'intervalo_max': high,
        'similar':       similar,
    })

    with out:
        clear_output(wait=True)
        sep = '─' * 52
        print(sep)
        print(f'  💰  Preço estimado      R$ {preco:>13,.0f}')
        print(f'  📉  Intervalo mínimo    R$ {low:>13,.0f}  (−15%)')
        print(f'  📈  Intervalo máximo    R$ {high:>13,.0f}  (+15%)')
        print(sep)
        print(f'  🏘️   Mediana {w_bairro.value[:18]:<18}  R$ {med_bairro:>10,.0f}')
        print(f'  📊  Estimativa vs. bairro   {sinal} {abs(delta):.1f}%')
        print(f'  🔢  Imóveis similares na base    {n_imoveis:,}')
        if sim_preco is not None:
            print(sep)
            print(f'  🔍  Imóvel real mais parecido encontrado:')
            print(f'      Bairro: {sim_bairro}  |  Tipo: {sim_tipo}')
            print(f'      Área: {sim_area:.0f} m²  |  Quartos: {sim_quartos}  |  Vagas: {sim_vagas}')
            print(f'      Preço real anunciado:  R$ {sim_preco:>12,.0f}')
            print(f'      Diferença vs. real:    {sim_sinal} {abs(sim_diff):.1f}%')
        print(sep)

# ── Botão de salvar ───────────────────────────────────────────────
btn_salvar = widgets.Button(
    description='💾 Salvar previsão',
    button_style='success',
    layout=widgets.Layout(width='200px', margin='12px 0 0 0')
)

def salvar(b):
    if not estado_atual:
        return

    similar = estado_atual.get('similar')

    linha = {
        # Entrada do usuário
        'bairro_simulado':      estado_atual['bairro'],
        'tipo_simulado':        estado_atual['tipo_imovel'],
        'area_simulada_m2':     estado_atual['area_util'],
        'quartos_simulados':    estado_atual['quartos'],
        'banheiros_simulados':  estado_atual['banheiros'],
        'suites_simuladas':     estado_atual['suites'],
        'vagas_simuladas':      estado_atual['vagas_garagem'],
        # Previsão
        'preco_previsto_R$':    round(estado_atual['preco_previsto'], 0),
        'intervalo_min_R$':     round(estado_atual['intervalo_min'], 0),
        'intervalo_max_R$':     round(estado_atual['intervalo_max'], 0),
        # Imóvel real comparável
        'real_bairro':          similar['bairro']        if similar is not None else '-',
        'real_tipo':            similar['tipo_imovel']   if similar is not None else '-',
        'real_area_m2':         similar['area_util']     if similar is not None else '-',
        'real_quartos':         int(similar['quartos'])  if similar is not None else '-',
        'real_preco_R$':        round(similar['preco_venda'], 0) if similar is not None else '-',
        'diferenca_pct':        round((estado_atual['preco_previsto'] - similar['preco_venda']) /
                                      similar['preco_venda'] * 100, 1) if similar is not None else '-',
    }

    caminho = '../data/simulacoes_salvas.csv'
    novo_df = pd.DataFrame([linha])

    if os.path.exists(caminho):
        existente = pd.read_csv(caminho, encoding='utf-8-sig')
        resultado = pd.concat([existente, novo_df], ignore_index=True)
    else:
        resultado = novo_df

    resultado.to_csv(caminho, index=False, encoding='utf-8-sig')

    with out_salvo:
        clear_output(wait=True)
        print(f'✅ Salvo! Total de simulações registradas: {len(resultado)}')
        print(f'   Arquivo: {caminho}')

btn_salvar.on_click(salvar)

# Conecta widgets à função de previsão
for w in [w_bairro, w_tipo, w_area, w_quartos, w_banheiros, w_suites, w_vagas]:
    w.observe(prever, names='value')

# ── Layout ────────────────────────────────────────────────────────
titulo = widgets.HTML(
    '<h3 style="font-family:sans-serif;margin:8px 0">'
    '🏙️ Simulador de Preço — Imóveis SP</h3>'
    '<p style="font-family:sans-serif;color:#555;margin:0 0 12px">'
    'Ajuste os parâmetros e clique em Salvar para registrar a simulação.</p>'
)

col_esq = widgets.VBox([w_bairro, w_tipo, w_area])
col_dir = widgets.VBox([w_quartos, w_banheiros, w_suites, w_vagas])
painel  = widgets.HBox([col_esq, col_dir], layout=widgets.Layout(gap='32px'))

display(widgets.VBox([titulo, painel, out, btn_salvar, out_salvo]))
prever()
```
